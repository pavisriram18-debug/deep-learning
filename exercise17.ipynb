```python
# Gradient Descent and Gradient Descent with Momentum

# Function : y = x^2
# Gradient : dy/dx = 2x

# Initial Values
x = 2
gradient = 2 * x

w = 0.5          # Initial Weight
alpha = 0.5      # Learning Rate

print("Gradient =", gradient)

# -------------------------------
# Gradient Descent
# W(t+1) = W(t) - alpha * Gradient
# -------------------------------

w_gd = w - alpha * gradient

print("\nGradient Descent")
print("Updated Weight =", w_gd)

# -------------------------------
# Gradient Descent with Momentum
# V(t+1) = mu*V(t) - alpha*Gradient
# W(t+1) = W(t) + V(t+1)
# -------------------------------

mu = 0.1
v = 2

v_new = mu * v - alpha * gradient
w_new = w + v_new

print("\nGradient Descent with Momentum")
print("Updated Velocity =", v_new)
print("Updated Weight =", w_new)
```
